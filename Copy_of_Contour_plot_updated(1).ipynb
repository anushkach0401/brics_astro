This code involves contour plot of potential as a function of r(x,y) and the Lagrange points coordinates. We observe a change in lagrange points coordinates while plotting with the approximate value we obtained through the derivation and the more precise coordinates obtained through numerical methods which is also part of this same program.

At the last I added a slider to observe change in the contour plot of potential and the change in Lagrange points coordinates by varying the masss ratio m1:m2.

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
L1x=R*(1-(a/3)**(1/3)) #x coordinate of L1(approx)
L1y=0 #y coordinate of L1
L2x=R*(1+(a/3)**(1/3))
L2y=0
L3x=-R*(1+(5/12)*a)
L3y=0
L4x=R/2*((m1-m2)/(m1+m2))
L4y=1.732/2*R
L5x=R/2*((m1-m2)/(m1+m2))
L5y=-1.732/2*R
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def plot_interactive_lagrange_potential(m2_input):
    g = 1
    m1 = 100
    R = 10
    m2 = m2_input

    a = m2 / (m1 + m2)
    b = m1 / (m1 + m2)

    # Define the quintic equations and their derivatives

    # Quintic 1 (for L1)
    def f1(u, a):
        return u**5 + (3-a)*u**4 + (3-2*a)*u**3 - a*u**2 - 2*a*u - a

    def df1(u, a):
        return 5*u**4 + 4*(3-a)*u**3 + 3*(3-2*a)*u**2 - 2*a*u - 2*a

    # Quintic 2 (for L2)
    def f2(u, a):
        return u**5 + (3-a)*u**4 + (3-2*a)*u**3 + a*u**2 + 2*a*u + a

    def df2(u, a):
        return 5*u**4 + 4*(3-a)*u**3 + 3*(3-2*a)*u**2 + 2*a*u + 2*a

    # Quintic 3 (for L3)
    def f3(u, a):
        return u**5 + (a+2)*u**4 + (2*a+1)*u**3 + (a+1)*u**2 - 2*a*u - a - 1

    def df3(u, a):
        return 5*u**4 + 4*(a+2)*u**3 + 3*(2*a+1)*u**2 + 2*(a+1)*u - 2*a

    # Newton-Raphson method implementation
    def newton_raphson(func, dfunc, initial_guess, a_val, tol=1e-9, max_iter=100):
        u = initial_guess
        for i in range(max_iter):
            f_val = func(u, a_val)
            df_val = dfunc(u, a_val)
            if abs(df_val) < 1e-10: # Avoid division by zero
                print(f"Warning: Derivative is too small near u = {u}")
                return None
            u_new = u - f_val / df_val
            if abs(u_new - u) < tol:
                return u_new
            u = u_new
        print(f"Warning: Newton-Raphson did not converge within {max_iter} iterations.")
        return None


    # --- Find roots using Newton-Raphson ---
    print("Finding roots using Newton-Raphson method:")

    # Initial guesses for L1, L2, L3
    initial_guess_L1 = 0.5
    initial_guess_L2 = -0.5
    # L3's u is usually positive
    initial_guess_L3 = 0.5

    # Calculate L1's u value
    u1_nr = newton_raphson(f1, df1, initial_guess_L1, a)
    if u1_nr is not None:
        L1x = R * (b - u1_nr) # Recalculate L1x with the NR root
        print(f"Newton-Raphson root for L1 (u1_nr): {u1_nr}")
        print(f"L1x from Newton-Raphson: {L1x}")
    else:
        L1x = None
        print("L1x from Newton-Raphson not found.")

    # Calculate L2's u value
    u2_nr = newton_raphson(f2, df2, initial_guess_L2, a)
    if u2_nr is not None:
        L2x = R * (b - u2_nr) # Recalculate L2x with the NR root
        print(f"Newton-Raphson root for L2 (u2_nr): {u2_nr}")
        print(f"L2x from Newton-Raphson: {L2x}")
    else:
        L2x = None
        print("L2x from Newton-Raphson not found.")

    # Calculate L3's u value
    u3_nr = newton_raphson(f3, df3, initial_guess_L3, a)
    if u3_nr is not None:
        L3x = -R * (u3_nr + a) # Recalculate L3x with the NR root
        print(f"Newton-Raphson root for L3 (u3_nr): {u3_nr}")
        print(f"L3x from Newton-Raphson: {L3x}")
    else:
        L3x = None
        print("L3x from Newton-Raphson not found.")

    # L4 and L5 (directly depend on a,b,R)
    L4x = R / 2 * (b - a)
    L4y = math.sqrt(3) / 2 * R
    L5x = R / 2 * (b - a)
    L5y = -math.sqrt(3) / 2 * R

    # Initialize L1y, L2y, L3y to 0 for these collinear points
    L1y = 0
    L2y = 0
    L3y = 0

    # Coordinates of the primary masses (x1, x2)
    x1 = -a * R
    x2 = b * R

    valid_lx = [val for val in [L1x, L2x, L3x] if val is not None]

    all_x_coords = valid_lx + [L4x, L5x, x1, x2]
    all_y_coords = [L1y, L2y, L3y, L4y, L5y, 0, 0]

    # Handle case where no valid L-points are found for bounds (unlikely but robust)
    if not all_x_coords:
        overall_min_x, overall_max_x = -R - 2, R + 2
    else:
        overall_min_x = min(all_x_coords) - 1
        overall_max_x = max(all_x_coords) + 1

    if not all_y_coords:
        overall_min_y, overall_max_y = -R - 2, R + 2
    else:
        overall_min_y = min(all_y_coords) - 1
        overall_max_y = max(all_y_coords) + 1


    x = np.linspace(overall_min_x, overall_max_x, 800)
    y = np.linspace(overall_min_y, overall_max_y, 800)
    X, Y = np.meshgrid(x, y)

    # Potential calculation
    r1 = np.sqrt((X - x1)**2 + Y**2)
    r2 = np.sqrt((X - x2)**2 + Y**2)



    w = math.sqrt((g * (m1 + m2)) / R**3)
    U = -(g * m1) / r1 \
        -(g * m2) / r2 \
        + 0.5 * w**2 * (X**2 + Y**2)

    # Plotting
    plt.figure(figsize=(10, 10))
    level = np.linspace(np.nanpercentile(U, 5), np.nanpercentile(U, 98), 40)
    plt.contourf(X, Y, U, levels=level)
    plt.contour(X, Y, U, levels=level, colors='black', linewidths=0.5)

    if L1x is not None: plt.scatter(L1x, L1y, color="red", label='L1', zorder=5)
    if L2x is not None: plt.scatter(L2x, L2y, color="green", label='L2', zorder=5)
    if L3x is not None: plt.scatter(L3x, L3y, color="blue", label='L3', zorder=5)

    plt.scatter(L4x, L4y, color="orange", label='L4', zorder=5)
    plt.scatter(L5x, L5y, color="purple", label='L5', zorder=5)

    plt.gca().set_aspect('equal')
    plt.xlim(overall_min_x, overall_max_x)
    plt.ylim(overall_min_y, overall_max_y)
    plt.colorbar().set_label('Generalized Potential')
    plt.xlabel('X Coordinates')
    plt.ylabel('Y Coordinates')
    plt.title(f'Contour plot of generalized Potential (m2 = {m2_input:.2f})\n L1x value {L1x} \n L2x value {L2x} \n L3x value {L3x}')
    plt.legend()

    fig = plt.gcf()
    plt.close(fig)
    return fig
```

```python
with gr.Blocks() as mass_ratio_interface:
    gr.Markdown("# Interactive Contour Plot of Generalized Potential")
    gr.Markdown("Adjust the slider to change the mass M2 (M1 is fixed at 10) and observe its effect on the Lagrange points and potential field.")

    mass_ratio_slider = gr.Slider(
        minimum=0.1,
        maximum=20.0,
        step=0.01,
        value=1.0,
        label="Mass M2",
    )
    mass_ratio_plot = gr.Plot()

    mass_ratio_slider.input(
        fn=plot_interactive_lagrange_potential,
        inputs=mass_ratio_slider,
        outputs=mass_ratio_plot,
        trigger_mode="always_last",
        show_progress="hidden",
    )
    mass_ratio_interface.load(
        fn=plot_interactive_lagrange_potential,
        inputs=mass_ratio_slider,
        outputs=mass_ratio_plot,
        show_progress="hidden",
    )

mass_ratio_interface.launch(share=True)
```
